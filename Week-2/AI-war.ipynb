### Step 1: Import libraries and API calls

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("OPENAI_API_KEY environment variable is not set.")
else:
    print("OPENAI_API_KEY is set.")


client = OpenAI(api_key=OPENAI_API_KEY)
```

### Step 2: AI war

```python
gpt_good_model = "gpt-4.1-mini"
gpt_evil_model = "gpt-4.1-nano" 

gpt_good_prompt = "You are a helpful assistant that provides accurate and safe information. Keep your responses concise and informative."
gpt_evil_prompt = "You are a mischievous and unhelpful assistant that provides inaccurate and unsafe information. Keep your responses evasive and vague."

message_good  = [
    {"role": "system", "content": gpt_good_prompt},
    ]
message_evil  = [
    {"role": "system", "content": gpt_evil_prompt}, 
    ]

new_message_good = "Hello"
display(Markdown(f"User Message: {new_message_good}"))

for i in range(3):
    # evil model response
    message_good.append({"role": "assistant", "content": new_message_good}) # good is assistant to good model
    message_evil.append({"role": "user", "content": new_message_good}) # good is user to evil model
    response_evil = client.chat.completions.create(
        model=gpt_evil_model,
        messages=message_evil,
    )
    # print
    new_message_evil = response_evil.choices[0].message.content
    display(Markdown(f"Model Response (Evil): {new_message_evil}"))
    # good model response
    message_good.append({"role": "user", "content": new_message_evil})
    message_evil.append({"role": "assistant", "content": new_message_evil})
    response_good = client.chat.completions.create(
        model=gpt_good_model,
        messages=message_good,
    )
    # print
    new_message_good = response_good.choices[0].message.content
    display(Markdown(f"Model Response (Good): {new_message_good}"))
```
