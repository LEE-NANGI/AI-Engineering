```python
import random
from openai import OpenAI

client = OpenAI()

def roll_dice(sides: int = 6) -> int:
    """Roll a dice with the given number of sides."""
    return random.randint(1, sides)
```

```python
target = 6
max_attempts = 20
```

```python
history = ""
iteration = 0

while iteration < max_attempts:
    iteration += 1
    instructions = f"""
    You are playing a dice game. You need roll a {target} to win.
    You've made the following rolls so far: 
    {history if history else "No rolls yet."}
    Based on what you have rolled so far, decide your next action:
    Option A: If you have achived the target, respond with exactly: DONE: <message describing how you achieved the target>
    Option B: If you have not achieved the target, respond with exactly: ROLL THE DICE
    """
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role": "user", "content": instructions},
            ]
    )
    result = response.choices[0].message.content.strip()

    if "DONE:" in result:
        print(result)
        break
    if "ROLL THE DICE" in result:
        roll = roll_dice()
        print(f"Rolled: {roll}\n")
        history += f"Roll {iteration}: {roll}\n"

print(f"Game ended after {iteration-1} iterations.")
```
